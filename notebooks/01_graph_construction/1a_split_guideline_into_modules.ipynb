```python
from pypdf import PdfReader, PdfWriter
import os



def extract_pdf_pages(input_pdf, output_pdf, pages):
    if not os.path.exists(input_pdf):
        print(f"Error: The file at {input_pdf} was not found.")
        return

    reader = PdfReader(input_pdf)
    writer = PdfWriter()

    for page_num in pages:
        if page_num < len(reader.pages):
            writer.add_page(reader.pages[page_num])
        else:
            print(f"Warning: Page {page_num + 1} is out of range.")

    with open(output_pdf, "wb") as out_file:
        writer.write(out_file)

    print(f"Successfully saved to Drive: {output_pdf}")

# 4. Run the extraction
extract_pdf_pages(input_path, output_path, pages_to_extract)
```
