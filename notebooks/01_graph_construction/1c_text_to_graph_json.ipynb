````python
import os
from langchain_core.documents import Document
from langchain_experimental.graph_transformers import LLMGraphTransformer
from google.colab import userdata

# 1. Setup LLM (Ensure GEMINI_API_KEY is in your Colab Secrets)
from langchain_google_genai import ChatGoogleGenerativeAI
os.environ["GOOGLE_API_KEY"] = userdata.get('GEMINI_API_KEY')

llm = ChatGoogleGenerativeAI(model="gemini-2.5-pro", temperature=0)

# 2. Initialize the Transformer
# You can restrict 'allowed_nodes' and 'allowed_relationships'
# to keep the graph clinical and clean.
llm_transformer = LLMGraphTransformer(llm=llm)

# 3. Load your .txt files into LangChain Documents
input_folder = "/content/"
documents = []

# Sort files to maintain page order
files = sorted(os.listdir(input_folder), key=lambda x: int(x.split('.')[0]) if x.split('.')[0].isdigit() else 0)

for filename in files:
    if filename.endswith(".txt"):
        file_path = os.path.join(input_folder, filename)
        with open(file_path, "r", encoding="utf-8") as f:
            text = f.read()
            # Create a Document object with metadata
            doc = Document(
                page_content=text,
                metadata={"source": filename, "page": filename.split('.')[0]}
            )
            documents.append(doc)

print(f"Loaded {len(documents)} documents. Starting graph transformation...")

# 4. Convert Documents to Graph Documents
# This step uses the LLM to extract entities and relationships
graph_documents = llm_transformer.convert_to_graph_documents(documents)

print(f"✅ Successfully created {len(graph_documents)} Graph Documents.")

# 5. Preview the first extraction
if graph_documents:
    print("\n--- Sample Extraction (Page 1) ---")
    print(f"Nodes: {graph_documents[0].nodes[:3]}")
    print(f"Relationships: {graph_documents[0].relationships[:3]}")
````

````python
import pandas as pd

def save_graph_documents_to_csv(graph_documents, nodes_file="nodes.csv", rels_file="relationships.csv"):
    node_data = []
    rel_data = []

    # Track unique nodes to avoid duplicates
    seen_nodes = set()

    for doc in graph_documents:
        # 1. Process Nodes
        for node in doc.nodes:
            if node.id not in seen_nodes:
                node_data.append({
                    "id": node.id,
                    "label": node.type,
                    "name": node.id  # Often the ID is the primary name
                })
                seen_nodes.add(node.id)

        # 2. Process Relationships
        for rel in doc.relationships:
            rel_data.append({
                "source": rel.source.id,
                "target": rel.target.id,
                "type": rel.type
            })

    # Create DataFrames
    nodes_df = pd.DataFrame(node_data)
    rels_df = pd.DataFrame(rel_data)

    # Save to CSV
    nodes_df.to_csv(nodes_file, index=False)
    rels_df.to_csv(rels_file, index=False)

    print(f"✅ Created {nodes_file} ({len(nodes_df)} nodes)")
    print(f"✅ Created {rels_file} ({len(rels_df)} relationships)")

# Execute the conversion
save_graph_documents_to_csv(graph_documents)

# Download the files from Colab
from google.colab import files
files.download("nodes.csv")
files.download("relationships.csv")
````

````python
import json
import os
from neo4j import GraphDatabase

# 1. Setup Connection (Replace with your Aura details or local host)
# If using Aura, the URI looks like: "neo4j+s://xxxxxxxx.databases.neo4j.io"
URI = os.getenv('NEO4J_URI')
AUTH = (os.getenv('NEO4J_USERNAME'), os.getenv('NEO4J_PASSWORD'))



def import_clinical_graph(json_file):

    with open(json_file, 'r') as f:
        data = json.load(f)

    driver = GraphDatabase.driver(URI, auth=AUTH)

    with driver.session() as session:

        # ⚠️ optional — remove if you don’t want full wipe
        session.run("MATCH (n) DETACH DELETE n")

        # ------------------------
        # 1️⃣ Create Question & Action Nodes
        # ------------------------
        for node in data.get("nodes", []):
            session.run("""
                MERGE (n:Step {id: $id})
                SET n.text = $text,
                    n.type = $type,
                    n.phase = $phase
            """,
            id=node["id"],
            text=node.get("text", ""),
            type=node.get("type", ""),
            phase=node.get("phase", "")
            )

        # ------------------------
        # 2️⃣ Create Relationships (USE REAL TYPE)
        # ------------------------
        for rel in data.get("relationships", []):
            rel_type = rel["type"]

            session.run(f"""
                MATCH (a:Step {{id: $from_id}})
                MATCH (b:Step {{id: $to_id}})
                MERGE (a)-[:{rel_type}]->(b)
            """,
            from_id=rel["from"],
            to_id=rel["to"]
            )

        # ------------------------
        # 3️⃣ Create Diagnosis Nodes
        # ------------------------
        for diag in data.get("diagnosis", []):
            session.run("""
                MERGE (d:Diagnosis {id: $id})
                SET d.criteria = $criteria
            """,
            id=diag["id"],
            criteria=json.dumps(diag["criteria"])
            )

        # ------------------------
        # 4️⃣ Create Safety Trigger Nodes
        # ------------------------
        for safety in data.get("safety_triggers", []):
            session.run("""
                MERGE (s:SafetyTrigger {id: $id})
                SET s.condition = $condition,
                    s.risk = $risk
            """,
            id=safety["id"],
            condition=safety["condition"],
            risk=safety["risk"]
            )

        # ------------------------
        # 5️⃣ Create Management Nodes
        # ------------------------
        for m in data.get("management", []):
            session.run("""
                MERGE (m:Management {id: $id})
                SET m.applies_to = $applies_to,
                    m.interventions = $interventions
            """,
            id=m["id"],
            applies_to=m["applies_to"],
            interventions=m["interventions"]
            )

    driver.close()
    print("✨ Clinical Graph successfully built in Neo4j!")

# Run it
import_clinical_graph('mhgap_decision_graph.json')
````

````python
class ClinicalEngine:

    def __init__(self):
        self.driver = GraphDatabase.driver(URI, auth=AUTH)
        self.reset()

    def reset(self):
        self.state = {
            "current_node": "DEP_Q1",
            "answers": {},
            "core_symptom": False,
            "history_of_mania": False,
            "diagnosis": None,
            "finished": False
        }

    # ---------------------------
    # 1️⃣ Get current step
    # ---------------------------
    def get_current_step(self):
        with self.driver.session() as session:
            result = session.run("""
                MATCH (n:Step {id: $id})
                RETURN n.text AS text, n.type AS type
            """, id=self.state["current_node"])

            return result.single()

    # ---------------------------
    # 2️⃣ Move to next step
    # ---------------------------
    def move_next(self, answer):

        current = self.state["current_node"]
        self.state["answers"][current] = answer

        # track important clinical flags
        if current == "DEP_Q1" and answer == "yes":
            self.state["core_symptom"] = True

        if current == "DEP_Q6" and answer == "yes":
            self.state["history_of_mania"] = True

        rel_type = "NEXT_YES" if answer == "yes" else "NEXT_NO"

        with self.driver.session() as session:
            result = session.run(f"""
                MATCH (a:Step {{id: $current}})
                OPTIONAL MATCH (a)-[:{rel_type}]->(next)
                RETURN next.id AS next_id
            """, current=current)

            record = result.single()

            if record and record["next_id"]:
                self.state["current_node"] = record["next_id"]
                return "continue"
            else:
                self.evaluate_diagnosis()
                return "finished"

    # ---------------------------
    # 3️⃣ Evaluate Diagnosis
    # ---------------------------
    def evaluate_diagnosis(self):

        if self.state["core_symptom"]:

            if self.state["history_of_mania"]:
                self.state["diagnosis"] = "DEPRESSION_BIPOLAR"
            else:
                self.state["diagnosis"] = "DEPRESSION"
        else:
            self.state["diagnosis"] = "NO_DEPRESSION"

        self.state["finished"] = True

    # ---------------------------
    # 4️⃣ Get Management Plan
    # ---------------------------
    def get_management(self):

        diagnosis = self.state["diagnosis"]

        if diagnosis in ["DEPRESSION", "DEPRESSION_BIPOLAR"]:

            with self.driver.session() as session:
                result = session.run("""
                    MATCH (m:Management {applies_to: $diag})
                    RETURN m.interventions AS interventions
                """, diag=diagnosis)

                record = result.single()

                if record:
                    return record["interventions"]

        return ["No management required"]

    def close(self):
        self.driver.close()
````

````python
engine = ClinicalEngine()

while not engine.state["finished"]:

    step = engine.get_current_step()

    print("\nSystem:", step["text"])

    user_input = input("Answer (yes/no): ").strip().lower()

    result = engine.move_next(user_input)

    if result == "finished":
        break

print("\nDiagnosis:", engine.state["diagnosis"])

management = engine.get_management()

print("\nManagement Plan:")
for item in management:
    print("-", item)

engine.close()
````
